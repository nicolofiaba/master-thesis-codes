```python
import pickle
with open('/scratch/astro/nicolo.fiaba/trained_models/EL_ATT_UNET_hist_RGS000_0', "rb") as file_pi:
    hist_RGS000_0 = pickle.load(file_pi)
with open('/scratch/astro/nicolo.fiaba/trained_models/EL_ATT_UNET_hist_RGS000_minus4', "rb") as file_pi:
    hist_RGS000_minus4 = pickle.load(file_pi)
```

```python
import matplotlib.pyplot as plt
fig, axs = plt.subplots(1, 2, figsize=(10, 4))

plt.subplot(121)
plt.plot(hist_RGS000_0['loss'][0:], label='loss')
plt.plot(hist_RGS000_0['val_loss'][0:], label='validation loss')
plt.yscale('log')
plt.title("RGS000_0 model")
plt.legend()

plt.subplot(122)
plt.plot(hist_RGS000_minus4['loss'][0:], label='loss')
plt.plot(hist_RGS000_minus4['val_loss'][0:], label='validation loss')
plt.yscale('log')
plt.title("RGS000_minus4 model")
plt.legend()
```
